```python
import uuid
import numpy as np

class GenericObject:
    """
    Generic object data.
    """
    def __init__(self):
        self.id = uuid.uuid4()
        self.bb = (-1, -1, -1, -1)
        self.category= -1
        self.score = -1

class GenericImage:
    """
    Generic image data.
    """
    def __init__(self, filename):
        self.filename = filename
        self.tile = np.array([-1, -1, -1, -1])  # (pt_x, pt_y, pt_x+width, pt_y+height)
        self.objects = list([])

    def add_object(self, obj: GenericObject):
        self.objects.append(obj)
```

#### Training
Design and train a ffNN to deal with the “xview_recognition” classification task.

```python
import json

# Load database
# CAMBIO DE RUTAS: anotaciones bajo DATA_DIR.
# json_file = '../PROJECT/xview_recognition/xview_ann_train.json'
json_file = DATA_DIR / 'xview_ann_train.json'
with open(json_file) as ifs:
    json_data = json.load(ifs)
ifs.close()
```

```python
import numpy as np

counts = dict.fromkeys(categories.values(), 0)
anns = []
for json_img, json_ann in zip(json_data['images'].values(), json_data['annotations'].values()):
    image = GenericImage(json_img['filename'])
    image.tile = np.array([0, 0, json_img['width'], json_img['height']])
    obj = GenericObject()
    obj.bb = (int(json_ann['bbox'][0]), int(json_ann['bbox'][1]), int(json_ann['bbox'][2]), int(json_ann['bbox'][3]))
    obj.category = json_ann['category_id']
    # Contar los objetos de cada categoría
    counts[obj.category] += 1
    image.add_object(obj)
    anns.append(image)
print(counts)
```

```python
from sklearn.model_selection import train_test_split

anns_train, anns_valid = train_test_split(anns, test_size=0.1, random_state=1, shuffle=True)
print('Number of training images: ' + str(len(anns_train)))
print('Number of validation images: ' + str(len(anns_valid)))
```

```python
# EXPERIMENTO 11: combinar Dropout 0.1 y L2 1e-5 en las tres capas ocultas.
# Código sustituido del experimento 10:
# import hashlib
# import numpy as np
# from tensorflow.keras.models import Sequential
# from tensorflow.keras.layers import Input, Dense, Activation, Flatten, Dropout
#
# def build_model(seed, dropout_rate):
#     tf.keras.backend.clear_session()
#     tf.keras.utils.set_random_seed(seed)
#     model = Sequential([
#         Input(shape=(*IMAGE_SIZE, IMAGE_CHANNELS)),
#         Flatten(),
#         Dense(256, activation='relu'),
#         Dropout(dropout_rate),
#         Dense(128, activation='relu'),
#         Dropout(dropout_rate),
#         Dense(64, activation='relu'),
#         Dropout(dropout_rate),
#         Dense(len(categories)),
#         Activation('softmax'),
#     ])
#     return model
#
# def weights_fingerprint(model):
#     digest = hashlib.sha256()
#     for weights in model.get_weights():
#         digest.update(weights.tobytes())
#     return digest.hexdigest()
#
# CONTROL_DROPOUT = 0.0
# DROPOUT_RATE = 0.2
# model = build_model(SEED, DROPOUT_RATE)
# model.summary()

import hashlib
import numpy as np
from tensorflow.keras.regularizers import l2
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Activation, Flatten, Dropout

def build_model(seed, dropout_rate, l2_strength):
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)
    model = Sequential([
        Input(shape=(*IMAGE_SIZE, IMAGE_CHANNELS)),
        Flatten(),
        Dense(256, activation='relu', kernel_regularizer=l2(l2_strength)),
        Dropout(dropout_rate),
        Dense(128, activation='relu', kernel_regularizer=l2(l2_strength)),
        Dropout(dropout_rate),
        Dense(64, activation='relu', kernel_regularizer=l2(l2_strength)),
        Dropout(dropout_rate),
        Dense(len(categories)),
        Activation('softmax'),
    ])
    return model

def weights_fingerprint(model):
    digest = hashlib.sha256()
    for weights in model.get_weights():
        digest.update(weights.tobytes())
    return digest.hexdigest()

DROPOUT_RATE = 0.1
L2_STRENGTH = 1e-5
model = build_model(SEED, DROPOUT_RATE, L2_STRENGTH)
model.summary()
```

```python
# EXPERIMENTO 11: conservar SGD y registrar también entropía cruzada sin penalización.
# Código sustituido del experimento 10:
# from tensorflow.keras.optimizers import SGD
#
# def compile_model(model):
#     opt = SGD(learning_rate=1e-3, momentum=0.9, nesterov=True, clipnorm=1.0)
#     model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
#     return model
#
# model = compile_model(model)

from tensorflow.keras.optimizers import SGD

def compile_model(model):
    opt = SGD(learning_rate=1e-3, momentum=0.9, nesterov=True, clipnorm=1.0)
    model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy', tf.keras.metrics.CategoricalCrossentropy(name='cross_entropy')])
    return model

model = compile_model(model)
```

```python
# CAMBIO respecto a ffNN_example.ipynb: Sustituir el generador por una función de carga única de imágenes y etiquetas en memoria.
# Líneas sustituidas del original (fragmentos en su orden original):
# def generator_images(objs, batch_size, do_shuffle=False):
#     while True:
#         if do_shuffle:
#             np.random.shuffle(objs)
#         groups = [objs[i:i+batch_size] for i in range(0, len(objs), batch_size)]
#         for group in groups:
#             images, labels = [], []
#             for (filename, obj) in group:
#                 images.append(load_geoimage(filename))
#                 probabilities = np.zeros(len(categories))
#                 probabilities[list(categories.values()).index(obj.category)] = 1
#                 labels.append(probabilities)
#             images = np.array(images).astype(np.float32)
#             labels = np.array(labels).astype(np.float32)
#             yield images, labels

category_names = list(categories.values())
category_to_index = {name: index for index, name in enumerate(category_names)}

def prepare_images(objs):
    images = np.empty((len(objs), *IMAGE_SIZE, IMAGE_CHANNELS), dtype=np.float32)
    labels = np.zeros((len(objs), len(categories)), dtype=np.float32)
    for index, (filename, obj) in enumerate(objs):
        images[index] = load_geoimage(filename)
        labels[index, category_to_index[obj.category]] = 1.0
    return images, labels
```

```python
# CAMBIO respecto a ffNN_example.ipynb: Usar batch 8 y precargar entrenamiento y validación una sola vez.
# Líneas sustituidas del original (fragmentos en su orden original):
# batch_size = 16
# train_generator = generator_images(objs_train, batch_size, do_shuffle=True)
# valid_generator = generator_images(objs_valid, batch_size, do_shuffle=False)

objs_train = [(ann.filename, obj) for ann in anns_train for obj in ann.objects]
objs_valid = [(ann.filename, obj) for ann in anns_valid for obj in ann.objects]
batch_size = 8

x_train, y_train = prepare_images(objs_train)
x_valid, y_valid = prepare_images(objs_valid)
print(f'Entrenamiento: {x_train.shape}, rango [{x_train.min():.3f}, {x_train.max():.3f}]')
print(f'Validación: {x_valid.shape}, rango [{x_valid.min():.3f}, {x_valid.max():.3f}]')
print(f'Imágenes en memoria: {(x_train.nbytes + x_valid.nbytes) / 1024**2:.1f} MiB')
```

```python
# EXPERIMENTO 11: entrenar una combinación final, separar pérdida L2 y entropía cruzada y comparar con la referencia.
# Código sustituido del experimento 10:
# import json
# import shutil
# import numpy as np
#
# epochs = 200
# def evaluate_and_save(run_model, history, run_dir, lr_patience,
#                       stop_patience, initial_fingerprint, dropout_rate):
#     run_model.load_weights(str(run_dir / 'model.keras'))
#     train_metrics = run_model.evaluate(
#         x_train, y_train, batch_size=batch_size, verbose=0, return_dict=True)
#     valid_metrics = run_model.evaluate(
#         x_valid, y_valid, batch_size=batch_size, verbose=0, return_dict=True)
#     best_epoch = int(np.argmax(history.history['val_accuracy'])) + 1
#     metrics = {
#         'best_epoch': best_epoch,
#         'epochs_run': len(history.history['loss']),
#         'train_inference': {key: float(value) for key, value in train_metrics.items()},
#         'validation_inference': {key: float(value) for key, value in valid_metrics.items()},
#         'accuracy_gap_pp': float(100 * (train_metrics['accuracy'] - valid_metrics['accuracy'])),
#         'best_val_accuracy_fit': float(max(history.history['val_accuracy'])),
#     }
#     history_data = {key: [float(value) for value in values]
#                     for key, values in history.history.items()}
#     config = {
#         'seed': SEED, 'deterministic_ops': True,
#         'initial_weights_sha256': initial_fingerprint,
#         'image_size': list(IMAGE_SIZE), 'batch_size': batch_size,
#         'max_epochs': epochs, 'dropout': dropout_rate,
#         'hidden_units': [256, 128, 64],
#         'optimizer': {'name': 'SGD', 'learning_rate': 1e-3, 'momentum': 0.9,
#                       'nesterov': True, 'clipnorm': 1.0},
#         'split': {'test_size': 0.1, 'random_state': 1, 'shuffle': True},
#         'monitor': 'val_accuracy', 'reduce_lr_factor': 0.2,
#         'lr_patience': lr_patience, 'early_stop_patience': stop_patience,
#         'tensorflow_version': tf.__version__,
#     }
#     for filename, data in [('history.json', history_data),
#                            ('metrics.json', metrics), ('config.json', config)]:
#         with open(run_dir / filename, 'w') as outfile:
#             json.dump(data, outfile, indent=2)
#     print(f'Resultados: {run_dir}')
#     print(f'Mejor época: {best_epoch}; train (inferencia): {train_metrics["accuracy"]:.2%}; '
#           f'validación: {valid_metrics["accuracy"]:.2%}; '
#           f'brecha: {metrics["accuracy_gap_pp"]:.3f} puntos')
#     return metrics
#
# run_specs = [('control', 0.0), ('dropout_01', 0.1), ('dropout_02', 0.2)]
# run_results = {}
# initial_fingerprint = None
# for run_name, dropout_rate in run_specs:
#     run_dir = EXPERIMENT_DIR / run_name
#     run_dir.mkdir(parents=True, exist_ok=True)
#     print(f'EXPERIMENTO 10: {run_name}, Dropout={dropout_rate}, máximo={epochs}')
#     model = compile_model(build_model(SEED, dropout_rate))
#     fingerprint = weights_fingerprint(model)
#     if initial_fingerprint is None:
#         initial_fingerprint = fingerprint
#     assert fingerprint == initial_fingerprint, 'Inicializaciones diferentes'
#     h = model.fit(
#         x_train, y_train, batch_size=batch_size,
#         validation_data=(x_valid, y_valid), shuffle=True, epochs=epochs,
#         callbacks=make_callbacks(run_dir, LR_PATIENCE, STOP_PATIENCE), verbose=1,
#     )
#     metrics = evaluate_and_save(
#         model, h, run_dir, LR_PATIENCE, STOP_PATIENCE,
#         fingerprint, dropout_rate)
#     first_100 = h.history['val_accuracy'][:100]
#     metrics['best_val_accuracy_first_100'] = float(max(first_100))
#     metrics['best_epoch_first_100'] = int(np.argmax(first_100)) + 1
#     metrics['gain_after_100_pp'] = float(100 * (
#         metrics['best_val_accuracy_fit'] - max(first_100)))
#     with open(run_dir / 'metrics.json', 'w') as outfile:
#         json.dump(metrics, outfile, indent=2)
#     run_results[run_name] = {'dropout': dropout_rate, 'metrics': metrics}
#     with open(EXPERIMENT_DIR / 'comparison.json', 'w') as outfile:
#         json.dump({'seed': SEED, 'max_epochs': epochs, 'runs': run_results}, outfile, indent=2)
#
# control_accuracy = run_results['control']['metrics']['validation_inference']['accuracy']
# for result in run_results.values():
#     result['delta_vs_control_pp'] = float(100 * (
#         result['metrics']['validation_inference']['accuracy'] - control_accuracy))
#
# selected_run = max(run_results, key=lambda name: (
#     run_results[name]['metrics']['validation_inference']['accuracy'],
#     -run_results[name]['metrics']['validation_inference']['loss']))
# selected_rate = run_results[selected_run]['dropout']
# selected_dir = EXPERIMENT_DIR / selected_run
# shutil.copy2(selected_dir / 'model.keras', MODEL_PATH)
# model = tf.keras.models.load_model(str(MODEL_PATH))
# comparison = {
#     'seed': SEED, 'max_epochs': epochs, 'runs': run_results,
#     'selected_run': selected_run, 'selected_dropout': selected_rate,
#     'selection_rule': 'max val_accuracy; ties: min val_loss',
# }
# with open(EXPERIMENT_DIR / 'comparison.json', 'w') as outfile:
#     json.dump(comparison, outfile, indent=2)
# print('Comparación final con los mejores checkpoints en modo inferencia:')
# for name, result in run_results.items():
#     m = result['metrics']
#     print(f'{name}: train={m["train_inference"]["accuracy"]:.2%}, '
#           f'validación={m["validation_inference"]["accuracy"]:.2%}, '
#           f'brecha={m["accuracy_gap_pp"]:.3f} pp; '
#           f'ganancia después de 100 épocas={m["gain_after_100_pp"]:+.3f} pp')
# print(f'Modelo seleccionado para métricas por clase y test: {selected_run}')

import json
import numpy as np

epochs = 200
def evaluate_and_save(run_model, history, run_dir, lr_patience,
                      stop_patience, initial_fingerprint, dropout_rate):
    run_model = tf.keras.models.load_model(str(run_dir / 'model.keras'))
    train_metrics = run_model.evaluate(
        x_train, y_train, batch_size=batch_size, verbose=0, return_dict=True)
    valid_metrics = run_model.evaluate(
        x_valid, y_valid, batch_size=batch_size, verbose=0, return_dict=True)
    best_epoch = int(np.argmax(history.history['val_accuracy'])) + 1
    metrics = {
        'best_epoch': best_epoch,
        'epochs_run': len(history.history['loss']),
        'train_inference': {key: float(value) for key, value in train_metrics.items()},
        'validation_inference': {key: float(value) for key, value in valid_metrics.items()},
        'accuracy_gap_pp': float(100 * (train_metrics['accuracy'] - valid_metrics['accuracy'])),
        'best_val_accuracy_fit': float(max(history.history['val_accuracy'])),
        'regularization_loss': float(tf.add_n(run_model.losses).numpy()),
    }
    history_data = {key: [float(value) for value in values]
                    for key, values in history.history.items()}
    config = {
        'seed': SEED, 'deterministic_ops': True,
        'initial_weights_sha256': initial_fingerprint,
        'image_size': list(IMAGE_SIZE), 'batch_size': batch_size,
        'max_epochs': epochs, 'dropout': dropout_rate,
        'l2_strength': L2_STRENGTH,
        'hidden_units': [256, 128, 64],
        'optimizer': {'name': 'SGD', 'learning_rate': 1e-3, 'momentum': 0.9,
                      'nesterov': True, 'clipnorm': 1.0},
        'split': {'test_size': 0.1, 'random_state': 1, 'shuffle': True},
        'monitor': 'val_accuracy', 'reduce_lr_factor': 0.2,
        'lr_patience': lr_patience, 'early_stop_patience': stop_patience,
        'tensorflow_version': tf.__version__,
    }
    for filename, data in [('history.json', history_data),
                           ('metrics.json', metrics), ('config.json', config)]:
        with open(run_dir / filename, 'w') as outfile:
            json.dump(data, outfile, indent=2)
    print(f'Resultados: {run_dir}')
    print(f'Mejor época: {best_epoch}; train (inferencia): {train_metrics["accuracy"]:.2%}; '
          f'validación: {valid_metrics["accuracy"]:.2%}; '
          f'brecha: {metrics["accuracy_gap_pp"]:.3f} puntos')
    return metrics

# Una sola ejecución nueva; la referencia ya está entrenada y no se modifica.
model = compile_model(build_model(SEED, DROPOUT_RATE, L2_STRENGTH))
initial_fingerprint = weights_fingerprint(model)
reference_metrics = None
reference_status = 'no disponible'
config_path = REFERENCE_DIR / 'config.json'
metrics_path = REFERENCE_DIR / 'metrics.json'
if config_path.is_file() and metrics_path.is_file():
    with open(config_path) as source:
        reference_config = json.load(source)
    expected = {
        'seed': SEED, 'deterministic_ops': True,
        'initial_weights_sha256': initial_fingerprint,
        'image_size': list(IMAGE_SIZE), 'batch_size': batch_size,
        'max_epochs': epochs, 'dropout': DROPOUT_RATE,
        'hidden_units': [256, 128, 64],
        'optimizer': {'name': 'SGD', 'learning_rate': 1e-3, 'momentum': 0.9,
                      'nesterov': True, 'clipnorm': 1.0},
        'split': {'test_size': 0.1, 'random_state': 1, 'shuffle': True},
        'monitor': 'val_accuracy', 'reduce_lr_factor': 0.2,
        'lr_patience': LR_PATIENCE, 'early_stop_patience': STOP_PATIENCE,
        'tensorflow_version': tf.__version__,
    }
    mismatches = [key for key, value in expected.items()
                  if reference_config.get(key) != value]
    if reference_config.get('l2_strength', 0.0) != 0.0:
        mismatches.append('la referencia ya tiene L2')
    if mismatches:
        reference_status = 'no comparable: ' + ', '.join(mismatches)
    else:
        with open(metrics_path) as source:
            reference_metrics = json.load(source)
        reference_status = 'protocolo e inicialización compatibles; única regularización añadida: L2'
print('Referencia Dropout 0.1 sin L2:', reference_status)
print('EXPERIMENTO 11: Dropout 0.1 + L2 1e-5')
h = model.fit(
    x_train, y_train, batch_size=batch_size,
    validation_data=(x_valid, y_valid), shuffle=True, epochs=epochs,
    callbacks=make_callbacks(EXPERIMENT_DIR, LR_PATIENCE, STOP_PATIENCE), verbose=1,
)
experiment_metrics = evaluate_and_save(
    model, h, EXPERIMENT_DIR, LR_PATIENCE, STOP_PATIENCE,
    initial_fingerprint, DROPOUT_RATE)
model = tf.keras.models.load_model(str(MODEL_PATH))
best_idx = int(np.argmax(h.history['val_accuracy']))
best_value = np.max(h.history['val_accuracy'])
print('Best validation model: epoch ' + str(best_idx + 1) + ' - val_accuracy ' + str(best_value))
comparison = {
    'reference_status': reference_status, 'reference_dropout_01': reference_metrics,
    'experiment_11': experiment_metrics,
}
if reference_metrics is not None:
    comparison['validation_accuracy_delta_pp'] = float(100 * (
        experiment_metrics['validation_inference']['accuracy'] -
        reference_metrics['validation_inference']['accuracy']))
    # Sin L2, la pérdida de referencia es su entropía cruzada.
    comparison['validation_cross_entropy_delta'] = float(
        experiment_metrics['validation_inference']['cross_entropy'] -
        reference_metrics['validation_inference']['loss'])
    print(f'Diferencia de accuracy: {comparison["validation_accuracy_delta_pp"]:+.3f} pp')
    print(f'Diferencia de entropía cruzada sin L2: '
          f'{comparison["validation_cross_entropy_delta"]:+.6f}')
with open(EXPERIMENT_DIR / 'comparison.json', 'w') as outfile:
    json.dump(comparison, outfile, indent=2)
```

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

def draw_confusion_matrix(cm, categories):
    # Draw confusion matrix
    fig = plt.figure(figsize=[6.4*pow(len(categories), 0.5), 4.8*pow(len(categories), 0.5)])
    ax = fig.add_subplot(111)
    cm = cm.astype('float') / np.maximum(cm.sum(axis=1)[:, np.newaxis], np.finfo(np.float64).eps)
    im = ax.imshow(cm, interpolation='nearest', cmap=plt.colormaps['Blues'])
    ax.figure.colorbar(im, ax=ax)
    ax.set(xticks=np.arange(cm.shape[1]), yticks=np.arange(cm.shape[0]), xticklabels=list(categories.values()), yticklabels=list(categories.values()), ylabel='Annotation', xlabel='Prediction')
    # Rotate the tick labels and set their alignment
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
    # Loop over data dimensions and create text annotations
    thresh = cm.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, format(cm[i, j], '.2f'), ha="center", va="center", color="white" if cm[i, j] > thresh else "black", fontsize=int(20-pow(len(categories), 0.5)))
    fig.tight_layout()
    plt.show()
```

```python
import numpy as np

# Compute the accuracy
correct_samples_class = np.diag(cm).astype(float)
total_samples_class = np.sum(cm, axis=1).astype(float)
total_predicts_class = np.sum(cm, axis=0).astype(float)
print('Mean Accuracy: %.3f%%' % (np.sum(correct_samples_class) / np.sum(total_samples_class) * 100))
acc = correct_samples_class / np.maximum(total_samples_class, np.finfo(np.float64).eps)
print('Mean Recall: %.3f%%' % (acc.mean() * 100))
acc = correct_samples_class / np.maximum(total_predicts_class, np.finfo(np.float64).eps)
print('Mean Precision: %.3f%%' % (acc.mean() * 100))
for idx in range(len(categories)):
    # True/False Positives (TP/FP) refer to the number of predicted positives that were correct/incorrect.
    # True/False Negatives (TN/FN) refer to the number of predicted negatives that were correct/incorrect.
    tp = cm[idx, idx]
    fp = sum(cm[:, idx]) - tp
    fn = sum(cm[idx, :]) - tp
    tn = sum(np.delete(sum(cm) - cm[idx, :], idx))
    # True Positive Rate: proportion of real positive cases that were correctly predicted as positive.
    recall = tp / np.maximum(tp+fn, np.finfo(np.float64).eps)
    # Precision: proportion of predicted positive cases that were truly real positives.
    precision = tp / np.maximum(tp+fp, np.finfo(np.float64).eps)
    # True Negative Rate: proportion of real negative cases that were correctly predicted as negative.
    specificity = tn / np.maximum(tn+fp, np.finfo(np.float64).eps)
    # Dice coefficient refers to two times the intersection of two sets divided by the sum of their areas.
    # Dice = 2 |A∩B| / (|A|+|B|) = 2 TP / (2 TP + FP + FN)
    f1_score = 2 * ((precision * recall) / np.maximum(precision+recall, np.finfo(np.float64).eps))
    print('> %s: Recall: %.3f%% Precision: %.3f%% Specificity: %.3f%% Dice: %.3f%%' % (list(categories.values())[idx], recall*100, precision*100, specificity*100, f1_score*100))
```
